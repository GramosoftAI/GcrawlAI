### Step 2: Extract Clean Markdown from any URL
You can test using your free GcrawlAI API key from [gcrawlai.com/app](https://gcrawlai.com/app).

```python
import requests

API_KEY = "YOUR_GCRAWLAI_API_KEY"  # Get 50 free requests at https://gcrawlai.com/app
TARGET_URL = "https://news.ycombinator.com"

response = requests.post(
    "https://gcrawlai.com/api/scrape",
    headers={"Authorization": f"Bearer {API_KEY}"},
    json={
        "url": TARGET_URL,
        "render_js": True,
        "format": "markdown"
    }
)

data = response.json()
print("✅ Scrape Status:", response.status_code)
print("\n--- Extracted Markdown Sample ---\n")
print(data.get("markdown", "")[:1000])
```

### Step 3: Feeding into LangChain / LLM RAG Pipeline

```python
# The resulting Markdown is 82% smaller than raw HTML, saving massive token costs in LLM context windows.
markdown_content = data.get("markdown", "")
print(f"Total Characters: {len(markdown_content)}")
print(f"Estimated Tokens: ~{len(markdown_content) // 4}")
```
